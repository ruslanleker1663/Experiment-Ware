```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# 1. Membuat grid koordinat X dan Y (misal area 10km x 10km)
x = np.linspace(0, 10, 100)
y = np.linspace(0, 10, 100)
X, Y = np.meshgrid(x, y)

# 2. Mensimulasikan data ketinggian tanah (Z) dalam meter menggunakan NumPy
# Menggabungkan fungsi sin/cos untuk membuat efek bukit dan lembah
Z = 150 * (np.sin(X/2) * np.cos(Y/2)) + 200  # Ketinggian berkisar 50m - 350m

# 3. Membuat visualisasi berdampingan (2D Kontur & 3D Surface)
fig = plt.figure(figsize=(14, 6))

# --- Subplot 1: Peta Kontur 2D ---
ax1 = fig.add_subplot(121)
contour = ax1.contourf(X, Y, Z, levels=20, cmap='terrain') # cmap='terrain' sangat cocok untuk geografi
fig.colorbar(contour, ax=ax1, label='Ketinggian (Meter dpl)')
ax1.set_title('Peta Kontur Ketinggian Tanah (2D)')
ax1.set_xlabel('Koordinat X (km)')
ax1.set_ylabel('Koordinat Y (km)')

# --- Subplot 2: Model Relief 3D ---
ax2 = fig.add_subplot(122, projection='3d')
surface = ax2.plot_surface(X, Y, Z, cmap='terrain', edgecolor='none', alpha=0.9)
fig.colorbar(surface, ax=ax2, shrink=0.5, aspect=5, label='Ketinggian (Meter dpl)')
ax2.set_title('Model Visualisasi Relief Tanah (3D)')
ax2.set_xlabel('X (km)')
ax2.set_ylabel('Y (km)')
ax2.set_zlabel('Elevasi (m)')

# Mengatur sudut pandang kamera 3D agar relief terlihat jelas
ax2.view_init(elev=35, azim=-45)

plt.tight_layout()
plt.show()
```
